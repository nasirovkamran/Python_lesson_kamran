# Task 1

```python
score=int(input("Enter score(0-100): "))
if 0<=score<=100:
    print("Your score is ",end='')
    if score>=91:
        print("A")
    elif score>=81:
        print("B")
    elif score>=71:
        print("C")
    elif score>=61:
        print("D")
    elif score>=51:
        print("E")
    else:
        print("F")
else:
    print("Please enter the score correctly")
```

# Task 2

```python
n=int(input("Enter positive integer: "))
sm=0
cnt=0
for i in range(1,n+1):
    if i%3==0 and i%5==0:
        print(" 3 5")
        cnt+=1
    elif i%3==0:
        print("3")
        cnt+=1
    elif i%5==0:
        print("5")
    else:
        print(i)
    if i%2==0:
        sm+=i
print(f'The sum of all even numbers = {sm}')
print(f'{cnt} numbers were divisible by 3')
```
