```python
score=int(input("Enter your exam mark: "))
print("Your grade is ",end='')
if score>=90:
    print("A")
elif score>=75:
    print("B")
elif score>=50:
    print("C")
else:
    print("F")
```

```python
n=int(input("Enter a number: "))
for i in range(1,11):
    print(f'{n} * {i} = {n*i}')
```

```python
correct_password='kamran2008'
attempts=0
while attempts<3:
    password=input("Enter password: ")
    attempts+=1
    if password==correct_password:
        print("Password is true!")
        break
    else:
        print("Wrong password!")
        if attempts==3:
            print("Too many wrong attempts!")
```
