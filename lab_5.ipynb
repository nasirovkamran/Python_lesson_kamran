```python
data={
    'Math': 85,
    'Python': 90,
    'Database': 78,
    'Statistics': 92
}
```

```python
import pandas as pd
```

```python
s=pd.Series(data)
print(s)
```

```python
summ=0
for i in range(4):
    s.iloc[i]+=5
    summ+=s.iloc[i]
print(s)
print(f'sum = {summ}, average = {summ/4}')
```
